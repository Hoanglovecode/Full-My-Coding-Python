```python
import numpy as np
import matplotlib.pyplot as plt

# -------------------------
# Dữ liệu
# -------------------------
r2 = np.array([0.00, 0.01, 0.022, 0.04, 0.062])
I = np.array([0.005, 0.0092, 0.0155, 0.025, 0.0385])

# -------------------------
# Phương trình đường thẳng
# I = 0.5147*r² + 0.0041
# -------------------------
m = 0.5147
b = 0.0041

x = np.linspace(0, 0.062, 200)
y = m * x + b

# -------------------------
# Vẽ đồ thị
# -------------------------
plt.figure(figsize=(8,5))

# Các điểm thực nghiệm
plt.scatter(r2, I,
            s=100,
            color="#1f77b4",
            zorder=3)

# Đường hồi quy
plt.plot(x, y,
         linestyle=":",
         linewidth=2.5,
         color="black")

# Hiển thị phương trình
plt.text(0.0105, 0.032,
         r"$\bar{I}=0.5147\,r^2+0.0041$",
         fontsize=14)

# -------------------------
# Định dạng
# -------------------------
plt.xlim(0, 0.07)
plt.ylim(0.0, 0.04)

plt.xticks(np.arange(0, 0.071, 0.01))
plt.yticks(np.arange(0.005, 0.041, 0.005))

plt.xlabel(r"$r^2$ (m$^2$)", fontsize=14)
plt.ylabel(r"$\bar{I}$ (kg.m$^2$)", fontsize=14)

plt.grid(True, color="lightgray", linewidth=1)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# -----------------------------
# Data
# -----------------------------
x = np.array([0.000, 0.011, 0.022, 0.040, 0.062])
y = np.array([0.0050, 0.0092, 0.0155, 0.0250, 0.0385])

# Regression line
m = 0.5147
b = 0.0041

xx = np.linspace(0, 0.062, 500)
yy = m * xx + b

plt.rcParams['font.family'] = 'Calibri'     # Nếu máy không có sẽ tự chuyển font khác

fig, ax = plt.subplots(figsize=(9.3, 5.4), dpi=120)

# =============================
# Tiêu đề
# =============================
fig.suptitle(
    "ĐỒ THỊ BÁO CÁO THÍ NGHIỆM BÀI 3",
    fontsize=20,
    fontweight="bold",
    y=0.98
)

ax.set_title(
    "SỰ PHỤ THUỘC CỦA $\overline{I}$ VÀO $r^2$",
    fontsize=16,
    fontweight="bold",
    pad=12
)

# Grid giống Excel
ax.grid(
    True,
    color="#d0d0d0",
    linewidth=1
)

# Scatter
ax.scatter(
    x,
    y,
    s=120,
    color="#1f77b4",
    edgecolor="#1f77b4",
    zorder=3
)

# Trendline
ax.plot(
    xx,
    yy,
    color="black",
    linestyle=":",
    linewidth=2.5,
    zorder=2
)

# Equation
ax.text(
    0.010,
    0.032,
    r'$\bar{I}=0.5147\,r^2+0.0041$',
    fontsize=18
)

# Axis
ax.set_xlim(0, 0.07)
ax.set_ylim(0, 0.04)

ax.set_xticks(np.arange(0, 0.071, 0.01))
ax.set_yticks(np.arange(0.005, 0.041, 0.005))

ax.tick_params(labelsize=13)

ax.set_xlabel(r'$r^2$ (m$^2$)', fontsize=18)
ax.set_ylabel(r'$\bar{I}$ (kg.m$^2$)', fontsize=18)

# Border giống Excel
for s in ax.spines.values():
    s.set_linewidth(1)

# Để tiêu đề không bị chồng lên đồ thị
plt.tight_layout(rect=[0, 0, 1, 0.93])

plt.show()
```
