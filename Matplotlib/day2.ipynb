```python
import matplotlib.pyplot as plt
import numpy as np
#Bar chart: compare categories of data by representing each catefgory with a bar
```

```python
categories = np.array(["Grains", "Fruit", "Vegetables", "Protein", "Dairy","Sweets"])
values=np.array([4,3,2,5,3,1])
plt.bar(categories,values)
plt.title("Daily Consumprion")
plt.xlabel("Food")
plt.ylabel("Quantity")
plt.show()
```

```python
plt.barh(categories,values,color="#0509fc")
plt.title("Daily Consumprion")
plt.xlabel("Food")
plt.ylabel("Quantity")
plt.show()
```
