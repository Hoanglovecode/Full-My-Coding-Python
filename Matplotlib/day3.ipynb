```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

df = pd.read_csv("dataday3.csv")
print(df.to_string())
```

```python
print(df["Type1"].value_counts())
```

```python
type_count=df["Type1"].value_counts(ascending=True)
plt.barh(type_count.index,type_count.values,color="lightgreen",edgecolor="red")
plt.title("Number of pokemon by primary type")
plt.xlabel("Count")
plt.ylabel("Type")
plt.tight_layout()
plt.show()
```
