```python
import pandas as pd
```

```python
print(pd.__version__)
```

```python
data=[100.1,200,300]
series=pd.Series(data)
print(series)
```

```python
cost=[2000,2300,2500]
series=pd.Series(cost,index=["Department 1","Department 2","Department 3"])
print(series)
```

```python
print(series.loc["Department 1"])#label
print(series.iloc[0])#index
print(series.loc["Department 2"])
print(series.iloc[1])
print(series.loc["Department 3"])
print(series.iloc[2])
```

```python
series.loc["Department 1"]=3000
print(series)
```

```python
data=[100,102,104,200,202]
series=pd.Series(data,index=["a","b","c","d","e"])
print(series[series<200])
```

```python
calories={"Day 1":1730,"Day 2":1454,"Day 3":1500}
series=pd.Series(calories)
print(series)
```

```python
print(series.iloc[0])
print(series.loc["Day 2"])
```

```python
series.iloc[2]+=500
print(series.loc["Day 3"])
```

```python
print(series[series>1640])
```

```python
data={
    "Name":["Hoang","Braziler","Englander"],
    "Age":[19,25,54]
}
df=pd.DataFrame(data,index=["Stt1","Stt2","Stt3"])
print(df)
```

```python
print(df.loc["Stt2"])
```

```python
print(df.iloc[0])
```

```python
#Add a new column
df["Job"]=["Learn","Earn","Cook"]
print(df)
```

```python
new_row=pd.DataFrame([{"Name":"Italier","Age":50,"Job":"Clean"}],index=["Stt4"])
df=pd.concat([df,new_row])
print(df)
```

```python
new_rows=pd.DataFrame([{"Name":"Americaner","Age":19,"Job":"Worker"},
                       {"Name":"Francer","Age":20,"Job":"Worker"},
                       {"Name":"Spainer","Age":21,"Job":"Worker"}],
                       index=["Stt5","Stt6","Stt7"])
df=pd.concat([df,new_rows])
print(df)
```
