```python
#DATA CLEANING
#1.Delete Columns(Drop)
df=pd.read_csv("dataday2.csv")
df=df.drop(columns=["Legendary","No"])#Xóa cột No và legendary
print(df)
```

```python
#2.(Handle missing data)Loaị bỏ các thành phần NOT AVAILABLE cột Type2
df=df.dropna(subset="Type2")
print(df)
```

```python
df=pd.read_csv("dataday2.csv")
df=df.fillna({"Type2":"None"})# Thay thế các NaN thành None trong cột Type2
print(df)
```

```python
#3. Fix inconsistent value
#Thay thế các giá trị trong cột
df["Type1"]=df["Type1"].replace({"Grass":"GRASS",
                                 "Fire":"FIRE",
                                 "Water":"WATER",
                                 "Dragon":"DRAGON"})
print(df)
```

```python
#4. Standardize text
#Biến in HOA thành in thường
df["Name"]=df["Name"].str.lower()
print(df)
```

```python
#5. Fix data types
df["Legendary"]=df["Legendary"].astype(bool)
print(df)
```

```python
#Loại bỏ các thành phần giống nhau(Remove duplicate Values)
df=pd.read_json("dataday2.json")
print(df)
```

```python
df=df.drop_duplicates()
print(df)
```
