```python
class Company:
    class Employee:
        def __init__(self,name,position):
            self.name=name
            self.position=position
        def get_info(self):
            return f"{self.name} is {self.position} ?"
    def __init__(self,company_name):
        self.company_name=company_name
        self.employees=[]
    def add_employees(self,name,position):
        new_employee= self.Employee(name,position)
        self.employees.append(new_employee)
    def list_employees(self):
        return [employee.get_info() for employee in self.employees]
    
    @staticmethod
    def is_valid_position(position):
        valid_positions=["manager","chairperson","director"]
        return position in valid_positions
    
company1=Company("Al company")

company1.add_employees("hoang","director")
company1.add_employees("phong","cook")
company1.add_employees("long","chairperson")

for employee in company1.employees:
    print(employee.get_info())
    print(company1.is_valid_position(employee.position))
        
```
