```python
#Aggregation = Repesents a relationship where one objoect(whole)
#              contains references to one or more Independent objects (the parts)

class Lybrary:
    def __init__(self,name):
        self.name=name
        self.books=[]
    def add_book(self,book):
        self.books.append(book)

    def list_books(self):
        return [f"{book.title} by {book.author}" for book in self.books]

class Book:
    def __init__(self,title,author):
        self.title=title
        self.author=author

book1=Book("Toeic reading","Prep")
book2=Book("Toeic listening","Prep")
book3=Book("OOP for beginers","Lee Hoang")

lybrary=Lybrary("Hoang's library")

lybrary.add_book(book1)
lybrary.add_book(book2)
lybrary.add_book(book2)

print(lybrary.name)
for book in lybrary.list_books():
    print(book)

```
