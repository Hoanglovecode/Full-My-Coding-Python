```python
#Inheritance(Kế thừa)= Allows a class to inherit attributes and methods from another class
#              Helps with code reusability and extensibility
#              class Child(Parent)

class Animal:
    def __init__(self,name):
        self.name=name
        self.is_alive=True
    def speakk(self):
        print("Some sound")

    def eat(self):
        print(f"{self.name} is eating")
    def sleep(self):
        print(f"{self.name} is sleeping")

class Dog(Animal):
    def speak(self):
        super().speakk() # Call method parent
        print("Gau")
class Cat(Animal):
    def speak(self):
        print("Meow")
class Mouse(Animal):
    pass

dog=Dog("Bob")
cat=Cat("Tom")
mouse=Mouse("mickey")

print(dog.name)
print(dog.is_alive)
dog.eat()
dog.sleep()
dog.speak()

print("------------------------------")

print(cat.name)
print(cat.is_alive)
cat.eat()
cat.sleep()
cat.speak()
```

```python
#Multible inheritance = inherit from more than one parent class
#                       C(A,B)
#Multilevel inheritance = inherit from a parent which inherits from another parent
#                       C(A,B)<-B(A)<-A
class Animal:
    def __init__(self,name):
        self.name=name

    def eat(self):
        print(f"{self.name} is eating")

    def sleep(self):
        print(f"{self.name} is sleeping")

class Prey(Animal): #con mồi
    def flee(self): #bỏ chạy
        print(f"{self.name} is fleeing")

class Predator(Animal): #kẻ săn mồi
    def hunt(self): #săn
        print(f"{self.name} is hunting")

class Rabbit(Prey):
    pass

class Hawk(Predator):
    pass

class Fish(Prey,Predator): #Multible inheritance
    pass

rabbit=Rabbit("Thỏ")
hawl=Hawk("Chim ưng")
fish=Fish("Cá")

rabbit.flee()
hawl.hunt()
fish.hunt()
fish.flee()

rabbit.sleep()
fish.eat()
```

```python
#MRO: Method Resolution Order
class A:
    def hello(self):
        print("A")


class B(A):
    def hello(self):
        print("B")
        super().hello()


class C(A):
    def hello(self):
        print("C")
        super().hello()


class D(B, C):
    def hello(self):
        print("D")
        super().hello()
d = D()
d.hello()
```
