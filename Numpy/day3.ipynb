```python
x=np.array([i for i in range(10)])
print(x)
x
```

```python
np.where(x%2,'Even','Odd')
```

```python
x=np.arange(10)
x
```

```python
condlist=[x<5,x>5]
choicelist=[x**2,x**3]
np.select(condlist,choicelist,default=x)
```

if x < 5:
    dùng x**2

elif x > 5:
    dùng x**3

else:
    dùng default
    
np.select khác np.where ở chỗ là nó dùng được nhiều cặp đk

```python
arr=np.array([[1,2,5],[4,9,6],[7,8,3]])
arr
```

```python
#Lấy giá trị Max từng hàng
np.amax(arr,axis=1)
```

Tóm tắt
