Mẹo nhớ
Đối với broadcasting, hãy căn hai shape từ bên phải và kiểm tra từng chiều:
✅ Hai kích thước bằng nhau → được.
✅ Một trong hai bằng 1 → được (NumPy sẽ "kéo giãn" chiều đó).
❌ Khác nhau và không có số nào bằng 1 → lỗi.
Với ví dụ của bạn:
(1,4) * (4,1) → được vì cả hai chiều đều có ít nhất một giá trị là 1.
(2,4) * (4,1) → không được vì ở chiều đầu là 2 và 4, không bằng nhau và cũng không có 1.


```python
arr=np.array([[1,2,3,4]]) 
brr=np.array([[1],[2],[3],[4]]) 
print(arr.shape)
print(brr.shape)
print(arr*brr)
```

```python
np.multiply(arr,brr)
```

```python
np.divide(arr,brr)
```

```python
np.subtract(a,b)
```
