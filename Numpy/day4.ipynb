```python
rng=np.random.default_rng()
print(rng.integers(1,7))
```

```python
print(rng.integers(low=1,high=101))#k tính số 101
print(rng.integers(low=1,high=101,size=4))
print(rng.integers(low=1,high=101,size=(3,2)))
```
